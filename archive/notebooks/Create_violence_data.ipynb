```python
conflict_vars = [
    # Identifiers
    'country',
    'country_id',
    'year',
    'region',
    'govt_name',

    # Existence flags (any violence occurred, by type)
    'sb_exist',
    'sb_intrastate_exist',
    'sb_interstate_exist',
    'ns_exist',
    'os_exist',

    # Government involvement (relevant to sanctions -> regime repression channel)
    'sb_intrastate_govt_inv_incomp',
    'sb_interstate_govt_inv_incomp',
    'os_govt_inv',
    'os_any_govt_inv',

    # Dyad/actor counts (fragmentation of conflict, separate from intensity)
    'sb_dyad_count',
    'sb_intrastate_dyad_count',
    'sb_interstate_dyad_count',
    'ns_dyad_count',
    'os_dyad_count',

    # Cumulative totals (all violence types combined), by victim type
    'cumulative_total_deaths_parties_in_orgvio',
    'cumulative_total_deaths_civilians_in_orgvio',
    'cumulative_total_deaths_unknown_in_orgvio',
    'cumulative_total_deaths_in_orgvio_best',

    # State-based violence, by victim type
    'sb_deaths_parties',
    'sb_deaths_civilians',
    'sb_deaths_unknown',
    'sb_total_deaths_best',

    # State-based intrastate, by victim type
    'sb_intrastate_deaths_parties',
    'sb_intrastate_deaths_civilians',
    'sb_intrastate_deaths_unknown',
    'sb_intrastate_deaths_best',

    # State-based interstate, by victim type
    'sb_interstate_deaths_parties',
    'sb_interstate_deaths_civilians',
    'sb_interstate_deaths_unknown',
    'sb_interstate_deaths_best',

    # Non-state violence, by victim type
    'ns_deaths_parties',
    'ns_deaths_civilians',
    'ns_deaths_unknown',
    'ns_total_deaths_best',

    # One-sided violence, by perpetrator
    'os_govt_killings_best',
    'os_any_govt_killings_best',
    'os_nsgroup_killings_best',
    'os_killings_unknown',
    'os_total_deaths_best',
]

conflict_to_merge = df[conflict_vars]
```

```python
base_conflict=pd.merge(base, conflict_to_merge, how='left',
    left_on=['year','gwcode'], right_on=['year',  'country_id'])

base_conflict
```

```python
# columns you don't want to fillna with 0 (they're strings/labels, not counts)
non_numeric_cols = ['country', 'region', 'govt_name', 'year', 'country_id']

# numeric columns to zero-fill
numeric_conflict_vars = [c for c in conflict_vars if c not in non_numeric_cols]

mask = base_conflict['year'] > pd.Timestamp('1988-01-01')
base_conflict.loc[mask, numeric_conflict_vars] = base_conflict.loc[mask, numeric_conflict_vars].fillna(0)
```

```python
base_conflict[numeric_conflict_vars]
```

```python
print(base_conflict['pwtpop'].describe())
```

```python
# columns that should NOT be divided by population (binary existence/involvement flags)
binary_flags = [
    'sb_exist', 'sb_intrastate_exist', 'sb_interstate_exist',
    'ns_exist', 'os_exist',
    'sb_intrastate_govt_inv_incomp', 'sb_interstate_govt_inv_incomp',
    'os_govt_inv', 'os_any_govt_inv',
]

# columns to actually convert to per-capita (counts and death totals)
vars_to_percap = [c for c in numeric_conflict_vars if c not in binary_flags]

# pwtpop is in millions, so:
# deaths / pwtpop           -> deaths per million
# deaths / pwtpop / 10      -> deaths per 100,000  (equivalent to (deaths / (pwtpop * 1_000_000)) * 100_000)

for col in vars_to_percap:
    base_conflict[f'{col}_pc'] = (
        base_conflict[col].fillna(0) / base_conflict['pwtpop'] / 10
    )
```
