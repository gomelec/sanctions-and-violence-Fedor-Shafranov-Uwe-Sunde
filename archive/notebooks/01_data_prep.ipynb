```python


NOTEBOOK_DIR = None

def _find_nb_dir(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "01_data_prep.ipynb").exists():
            return candidate
    return start

ROOT = Path(NOTEBOOK_DIR) if NOTEBOOK_DIR else _find_nb_dir(Path.cwd().resolve())
print("Notebook root:", ROOT)

RAW_DIR  = ROOT / "data" / "data_raw"
CLEAN_DIR = ROOT / "data" / "data_clean"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

df_dyadic = pd.read_stata(RAW_DIR / "GSDB_V4_dyadic.dta")
df_main   = pd.read_csv(RAW_DIR / "GSDB_V4.csv")
```
